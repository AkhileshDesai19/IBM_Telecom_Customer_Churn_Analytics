# 🧹 Data Cleaning

Before performing exploratory data analysis, the dataset must be cleaned and validated to ensure data quality.

The following cleaning tasks will be performed:

- Check for missing values
- Handle blank values
- Convert incorrect data types
- Check duplicate records
- Standardize column names
- Remove extra spaces
- Validate categorical values
- Export cleaned dataset

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# check missing values
df.isnull().sum()
```

### Observation

The dataset does not contain any standard NULL values.

However, the **TotalCharges** column is stored as text and contains blank strings, which must be handled before converting it to a numeric data type.

```python
#Blank Values in TotalCharges
(df["TotalCharges"] == " ").sum()
```

```python
# convert TotalCharges to Numeric

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors= "coerce")
```

```python
# Checking missing values again, now we will see blank values 11
df.isnull().sum()
```

### Observation

After converting **TotalCharges** to a numeric data type, 11 missing values were identified.

These records belong to newly joined customers with zero tenure, where total charges have not yet accumulated.

Since they represent only a very small percentage of the dataset, they will be removed.

```python
#removing missing values
df = df.dropna()
```

```python
#Verify
df.isnull().sum()
```

## Dupilcates check

```python
df.duplicated().sum()
```

### Observation

No duplicate records were found in the dataset.

## Check Data Types

```python
df.dtypes
```

```python
# Standardize Column Names

df.columns = (
    df.columns
      .str.strip()
      .str.replace(" ", "_")
      .str.replace("-", "_")
      .str.lower()
)

df.columns
```

```python
df.rename(
    columns={
        "customerid": "customer_id",
        "seniorcitizen": "senior_citizen",
        "phoneservice": "phone_service",
        "multiplelines": "multiple_lines",
        "internetservice": "internet_service",
        "onlinesecurity": "online_security",
        "onlinebackup": "online_backup",
        "deviceprotection": "device_protection",
        "techsupport": "tech_support",
        "streamingtv": "streaming_tv",
        "streamingmovies": "streaming_movies",
        "paperlessbilling": "paperless_billing",
        "paymentmethod": "payment_method",
        "monthlycharges": "monthly_charges",
        "totalcharges": "total_charges"
    },
    inplace=True
)

df.columns
```

```python
# Remove Leading/Trailing Spaces

object_columns = df.select_dtypes(include="object").columns

for col in object_columns:
    df[col] = df[col].str.strip()
```

```python
# Validate Categories

for col in object_columns:
    print(f"\n{'='*50}")
    print(col)
    print(df[col].unique())
```

```python
df.info()
```

## ✅ Data Cleaning Summary

The following cleaning tasks were completed:

- Checked missing values
- Converted TotalCharges to numeric
- Removed rows with missing TotalCharges
- Verified duplicate records
- Standardized column names
- Removed extra spaces
- Validated categorical values

The cleaned dataset is now ready for exploratory data analysis (EDA).

## Save clean Dataset

```python
# Save Clean Dataset

df.to_csv(
    "../data/cleaned/telco_customer_churn_cleaned.csv",
    index=False
)

print("✅ Cleaned dataset exported successfully.")
```
