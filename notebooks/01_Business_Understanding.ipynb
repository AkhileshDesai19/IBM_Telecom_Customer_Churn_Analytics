# 📡 Telecom Customer Churn Analytics

## End-to-End Data Analytics Project

### Project Overview

Customer churn is one of the biggest challenges faced by telecom companies. Losing existing customers leads to reduced revenue, increased customer acquisition costs, and lower profitability.

This project aims to analyze customer churn by combining traditional data analytics techniques with a modern ETL pipeline. The project follows the complete analytics lifecycle, including business understanding, data cleaning, exploratory data analysis (EDA), feature engineering, AWS S3 integration, PostgreSQL data warehousing (Bronze → Silver → Gold), SQL business analysis, and interactive Power BI dashboards.

The objective is to identify the major factors contributing to customer churn and provide actionable business recommendations that help improve customer retention.

# 🎯 Business Problem

A telecom company has observed an increasing number of customers discontinuing its services. Customer churn directly impacts revenue, customer lifetime value, and long-term business growth.

The company wants to better understand customer behavior and identify the key reasons why customers leave. By analyzing customer demographics, subscribed services, billing information, contract details, and payment methods, the business aims to discover patterns associated with customer churn.

The insights generated from this analysis will help management design strategies to improve customer retention and reduce revenue loss.

# 🎯 Business Objectives

The primary objectives of this project are:

- Understand customer churn behavior.
- Identify the major factors influencing customer churn.
- Analyze customer demographics and service usage.
- Measure customer retention and churn rate.
- Analyze revenue generated by different customer segments.
- Identify high-risk customer groups.
- Build business-ready KPIs for management.
- Develop interactive Power BI dashboards.
- Provide actionable business recommendations for improving customer retention.

# ❓ Business Questions

This project aims to answer the following business questions:

1. What is the overall customer churn rate?

2. Which contract type experiences the highest customer churn?

3. Which payment method is associated with the highest churn?

4. Does internet service type influence customer churn?

5. Which customer segment contributes the highest revenue?

6. Does customer tenure affect churn?

7. How do monthly charges influence customer churn?

8. Which value-added services reduce customer churn?

9. Which customer segments should the company focus on retaining?

10. What business strategies can help reduce customer churn?

# 📊 Key Performance Indicators (KPIs)

# 📂 Dataset Understanding

This project uses the IBM Telco Customer Churn Dataset.

The dataset contains customer demographic information, subscribed telecom services, contract details, billing information, payment methods, and churn status.

The target variable is **Churn**, indicating whether a customer has left the company.

Each row represents one customer, while each column represents a customer attribute or subscribed service.
